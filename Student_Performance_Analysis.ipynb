```python
df['total score'] = df['math score'] + df['reading score'] + df['writing score']

df['average score'] = df['total score'] / 3

df[['math score', 'reading score', 'writing score', 'total score', 'average score']].head()
```

```python
top_10 = df[['math score', 'reading score', 'writing score', 'total score', 'average score']].sort_values(
    by='total score', ascending=False
).head(10)

top_10
```

```python
bottom_10 = df[['math score', 'reading score', 'writing score', 'total score', 'average score']].sort_values(
    by='total score', ascending=True
).head(10)

bottom_10
```

```python
df[['math score', 'reading score', 'writing score']].mean()
```

```python
subject_averages = df[['math score', 'reading score', 'writing score']].mean()

print("Highest Average Score:", subject_averages.idxmax())
print("Highest Average:", subject_averages.max())

print("Lowest Average Score:", subject_averages.idxmin())
print("Lowest Average:", subject_averages.min())
```

```python
passed = df[
    (df['math score'] >= 50) &
    (df['reading score'] >= 50) &
    (df['writing score'] >= 50)
]

failed = df[
    (df['math score'] < 50) |
    (df['reading score'] < 50) |
    (df['writing score'] < 50)
]

print("Students Passed:", len(passed))
print("Students Failed:", len(failed))
```

```python
import matplotlib.pyplot as plt

labels = ['Passed', 'Failed']
values = [len(passed), len(failed)]

plt.bar(labels, values)
plt.title('Pass vs Fail Students')
plt.xlabel('Result')
plt.ylabel('Number of Students')
plt.show()
```

```python
df['result'] = df['average score'].apply(
    lambda x: 'Passed' if x >= 40 else 'Failed'
)

gender_result = pd.crosstab(df['gender'], df['result'])

print(gender_result)

gender_result.plot(kind='bar', figsize=(8,5))

plt.title('Pass vs Fail Students by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.legend(title='Result')
plt.show()
```

```python
print(df['result'].value_counts())
```

```python
prep_result = pd.crosstab(
    df['test preparation course'],
    df['result']
)

print(prep_result)

prep_result.plot(kind='bar', figsize=(8,5))

plt.title('Pass vs Fail Students by Test Preparation Course')
plt.xlabel('Test Preparation Course')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.legend(title='Result')
plt.show()
```

```python
lunch_result = pd.crosstab(
    df['lunch'],
    df['result']
)

print(lunch_result)

lunch_result.plot(kind='bar', figsize=(8,5))

plt.title('Pass vs Fail Students by Lunch Type')
plt.xlabel('Lunch Type')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.legend(title='Result')
plt.show()
```

```python
parent_result = pd.crosstab(
    df['parental level of education'],
    df['result']
)

print(parent_result)

parent_result.plot(kind='bar', figsize=(10,5))

plt.title('Pass vs Fail Students by Parental Education')
plt.xlabel('Parental Level of Education')
plt.ylabel('Number of Students')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Result')
plt.tight_layout()
plt.show()
```

```python
race_result = pd.crosstab(df['race/ethnicity'], df['result'])

print(race_result)

race_result.plot(kind='bar', figsize=(8,5))

plt.title('Pass vs Fail Students by Race/Ethnicity')
plt.xlabel('Race/Ethnicity')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.legend(title='Result')
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['total score'], bins=10)

plt.title('Distribution of Total Scores')
plt.xlabel('Total Score')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['math score'], bins=10, alpha=0.6, label='Math')
plt.hist(df['reading score'], bins=10, alpha=0.6, label='Reading')
plt.hist(df['writing score'], bins=10, alpha=0.6, label='Writing')

plt.title('Distribution of Subject Scores')
plt.xlabel('Score')
plt.ylabel('Number of Students')
plt.legend()

plt.show()
```

## Final Key Findings

1. The overall pass rate is high, with 812 students passing and 188 students failing.

2. Reading has the highest average score (69.169), while Mathematics has the lowest average score (66.089).

3. The highest-performing students achieved scores close to the maximum total score of 300.

4. Students who completed the test preparation course generally had a higher pass rate than students who did not take the course.

5. Standard lunch students had a higher number of passed students compared with free/reduced lunch students.

6. Pass and fail outcomes were compared across gender, race/ethnicity, and parental education levels.

7. Most students' total scores were concentrated in the middle-to-high score range.

8. The subject score distributions show that most students scored between approximately 50 and 90 in each subject.

9. Correlation analysis shows a positive relationship between Mathematics, Reading, and Writing scores.

## Conclusion

The analysis shows that student performance is generally good, with the majority of students passing. Reading had the highest average performance, while Mathematics had the lowest average. Test preparation and other demographic factors were also compared to understand differences in student outcomes. Overall, the analysis provides useful insights into the factors associated with student academic performance.
