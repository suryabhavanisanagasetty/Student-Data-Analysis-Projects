```python
print("Mean:")
print(df.mean(numeric_only=True))

print("\nMedian:")
print(df.median(numeric_only=True))

print("\nMode:")
print(df.mode(numeric_only=True).iloc[0])
```

```python
correlation_matrix = df.corr(numeric_only=True)

print(correlation_matrix)
```

```python
plt.figure(figsize=(12, 8))

plt.imshow(correlation_matrix, cmap="coolwarm", interpolation="none")
plt.colorbar()

plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=90)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

plt.title("Correlation Heatmap of Wine Quality Dataset")
plt.tight_layout()
plt.show()
```

```python
high_correlation = correlation_matrix[
    (correlation_matrix.abs() >= 0.5) &
    (correlation_matrix.abs() < 1.0)
]

print("Highly Correlated Features:")
print(high_correlation.dropna(how="all").dropna(axis=1, how="all"))
```

```python
plt.figure(figsize=(14, 8))

df.boxplot()

plt.title("Outlier Detection in Wine Quality Dataset")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower=lower_limit, upper=upper_limit)

print("Outliers treated successfully!")
```

```python
high_quality_wines = df[df["quality"] >= 7]

print("High Quality Wines:")
print(high_quality_wines.head())

print("\nNumber of high quality wines:", len(high_quality_wines))
```

```python
sorted_wines = df.sort_values(by="quality", ascending=False)

print("Wines sorted by quality:")
print(sorted_wines.head(10))
```

```python
quality_counts = df["quality"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(quality_counts.index, quality_counts.values)

plt.xlabel("Wine Quality")
plt.ylabel("Number of Wines")
plt.title("Wine Quality Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["alcohol"], bins=10)

plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.title("Distribution of Alcohol Content")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["alcohol"], df["quality"])

plt.xlabel("Alcohol")
plt.ylabel("Wine Quality")
plt.title("Alcohol Content vs Wine Quality")

plt.show()
```

```python
average_alcohol = df.groupby("quality")["alcohol"].mean()

plt.figure(figsize=(8, 5))

plt.plot(average_alcohol.index, average_alcohol.values, marker="o")

plt.xlabel("Wine Quality")
plt.ylabel("Average Alcohol Content")
plt.title("Average Alcohol Content by Wine Quality")

plt.show()
```

```python
print("Average Alcohol Content by Wine Quality:")
print(average_alcohol)

print("\nHighest average alcohol content is found in quality:",
      average_alcohol.idxmax())

print("Lowest average alcohol content is found in quality:",
      average_alcohol.idxmin())
```
