```python
df = pd.read_csv("Superstore.csv", encoding="latin1")

print("Dataset loaded successfully!")
```

```python
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
df.head()
```

```python
print("Column Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())

print("\nAverage Sales:", df["Sales"].mean())
print("Average Profit:", df["Profit"].mean())

print("\nMaximum Sales:", df["Sales"].max())
print("Maximum Profit:", df["Profit"].max())

print("\nMinimum Sales:", df["Sales"].min())
print("Minimum Profit:", df["Profit"].min())
```

```python
region_performance = df.groupby("Region")[["Sales", "Profit"]].sum().sort_values(
    by="Sales", ascending=False
)

print("Region-wise Sales and Profit:")
print(region_performance)
```

```python
category_performance = df.groupby("Category")[["Sales", "Profit"]].sum().sort_values(
    by="Sales", ascending=False
)

print("Category-wise Sales and Profit:")
print(category_performance)
```

```python
high_sales = df[df["Sales"] > 500]

print("Orders with Sales greater than 500:")
print(high_sales.head())

print("\nNumber of high-sales orders:", len(high_sales))
```

```python
sorted_sales = df.sort_values(by="Sales", ascending=False)

print("Top 10 Orders by Sales:")
print(sorted_sales[["Order ID", "Category", "Sales", "Profit"]].head(10))
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))

plt.bar(region_sales.index, region_sales.values)

plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.title("Region-wise Sales Performance")

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))

plt.bar(category_sales.index, category_sales.values)

plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.title("Category-wise Sales Performance")

plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Monthly Sales Trend")

plt.xticks(rotation=90)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=30)

plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Distribution of Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Sales"], df["Profit"], alpha=0.5)

plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")

plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)

print("Region-wise Profit:")
print(region_profit)

plt.figure(figsize=(8, 5))

plt.bar(region_profit.index, region_profit.values)

plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.title("Region-wise Profit Performance")

plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

print("Category-wise Profit:")
print(category_profit)

plt.figure(figsize=(8, 5))

plt.bar(category_profit.index, category_profit.values)

plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.title("Category-wise Profit Performance")

plt.show()
```

```python
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

print("Top 10 Products by Sales:")
print(top_products)
```

```python
plt.figure(figsize=(10, 6))

plt.barh(top_products.index[::-1], top_products.values[::-1])

plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.title("Top 10 Products by Sales")

plt.tight_layout()
plt.show()
```

```python
print("Average Sales:", df["Sales"].mean())
print("Average Profit:", df["Profit"].mean())
print("Total Quantity Sold:", df["Quantity"].sum())
print("Average Quantity per Order:", df["Quantity"].mean())
print("Average Discount:", df["Discount"].mean())
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Overall Profit Margin:", profit_margin, "%")
```
